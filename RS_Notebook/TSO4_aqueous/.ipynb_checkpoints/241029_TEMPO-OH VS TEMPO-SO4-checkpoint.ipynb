```python
from pyspecdata import find_file, plot
from matplotlib.pyplot import (
    text,
    axvline,
    figure,
    legend,
    savefig,
    rcParams,
    gca,
    show,
)
```

```python
# load file
d = {}  # dictionary of nddata
for k, fname in [
    # TEMPO-OH samples
    (
        "d_TEMPO_OH_50mM",
        "241029_TEMPO_OH_50mM.DSC",
        # Sample name: 50 mM TEMPO-OH in water
    ),
    (
        "d_TEMPO_OH_100uM_try1",
        "241029_TEMPO_OH_100uM.DSC"
        # Sample name: 100 uM TEMPO-OH in water trial 1
    ),
    (
        "d_TEMPO_OH_100uM_try2",
        "241029_TEMPO_OH_100uM_try2.DSC"
        # Sample name: 100 uM TEMPO-OH in water trial 2
    ),
    # TEMPO-SO4 samples
    (
        "d_TEMPO_SO4_50mM",
        "241029_TEMPO_SO4_50mM.DSC",
        # Sample name: 50 mM TEMPO-SO4 in water
    ),
    (
        "d_TEMPO_SO4_100uM_try1",
        "241029_TEMPO_SO4_100uM.DSC"
        # Sample name: 100 uM TEMPO-SO4 in water trial 1
    ),
    (
        "d_TEMPO_SO4_100uM_try2",
        "241029_TEMPO_SO4_100uM_try2.DSC"
        # Sample name: 100 uM TEMPO-SO4 in water trial 2
    ), 
    ## Compare and recheck TEMPO-SO4
    (
        "d_TEMPO_SO4_100uM_241217_old",
        "241217_100uM_TSO4_aqueous.DSC"
        # Sample name: 100 uM TEMPO-SO4 in water 241217 prepared from old 100 mM
    ), 
    (
        "d_TEMPO_SO4_100uM_241218_old",
        "241218_100uM_TSO4_aq_old20mM.DSC"
        # Sample name: 100 uM TEMPO-SO4 in water 241218 prepared from old 20 mM
    ), 
    (
        "d_TEMPO_SO4_NewSolid_100uM_241217",
        "241217_NewSolid_100uM_TSO4_aqueous.DSC"
        # Sample name: 100 uM TEMPO-SO4 prepared from Solid TSO4 241217
    ), 
    ## Re compare TEMPO-SO4 250109
    (
        "d_TEMPO_SO4_100uM_aged_241218_ESR_250109",
        "250109_100uM_TSO4_aq_old.DSC"
        # Sample name: 100 uM TEMPO-SO4 in water prepared at 241218 from old 20 mM and ESR recheck at 250109
    ), 
    (
        "d_TEMPO_SO4_100uM_NewSolid_241217_ESR_250109",
        "250109_100uM_TSO4_aq_NewSolid.DSC"
        # Sample name: 100 uM TEMPO-SO4 prepared from Solid TSO4 241217 and ESR recheck at 250109
    ),
    ### reprepare from solid 250110
    (
        "d_TEMPO_SO4_20mM_2ndNewSolid_250110",
        "250110_20mM_TSO4_aq_2ndNewSolid.DSC"
        # Sample name: 20 mM TEMPO-SO4 prepared from Solid TSO4 (2nd) 250110
    ),
    (
        "d_TEMPO_SO4_100uM_2ndNewSolid_250110",
        "250110_100uM_TSO4_aq_2ndNewSolid.DSC"
        # Sample name: 100 uM TEMPO-SO4 in water prepared from 20 mM 2nd New Solid 2nd at 250110
    ),
    
    # 2nd Synthesis
    (
        "d_TEMPO_SO4_20mM_2ndSynthesis",
        "250112_TEMPOSO4_20mM_2ndSynthesis.DSC"
        # Sample name: 20 mM TEMPO-SO4 in water prepared from 2nd Synthesis
    ),
    (
        "d_TEMPO_SO4_100uM_2ndSynthesis",
        "250112_TEMPOSO4_100uM_2ndSynthesis.DSC"
        # Sample name: 100 uM TEMPO-SO4 in water prepared from 2nd Synthesis diluting 20 mM
    ),
    (
        "d_TEMPO_OH_100uM_try1",
        "241029_TEMPO_OH_100uM.DSC"
        # Sample name: 100 uM TEMPO-OH in water trial 1
    ),
    # longer conversion time
        ## TEMPO-OH
    (
        "d_TEMPO_OH_100uM_CC12_250123",
        "250123_TEMPOL_100uM_CC12.DSC"
        # Sample name: 100 uM TEMPO-OH in water 250123
    ),
        ## TEMPO-SO4 reference
    (
        "d_TEMPO_SO4_100uM_CC12_JFreference_250122",
        "250122_100uM_TSO4_JF_reference.DSC"
        # Sample name: 100 uM TEMPO-SO4 in water 2nd Synthesis 250112
    ),
        ## TEMPO-SO4 aged
    (
        "d_TEMPO_SO4_100uM_CC12_aged_Prep241218",
        "250123_TSO4_100uM_aged_prep241218.DSC"
        # Sample name: 100 uM TEMPO-SO4 in water aged prepared 241218
    ),
    (
        "d_TEMPO_SO4_100uM_CC12_aged_Prep_fromSolid_241217",
        "250123_TSO4_100uM_aged_prep_FromSolid241217.DSC"
        # Sample name: 100 uM TEMPO-SO4 in water aged prepared from solid 241217
    ),
    (
        "d_TEMPO_SO4_100uM_CC12_aged_Prep_fromSolid_250110",
        "250123_TSO4_100uM_aged_prep_From2ndSolid250110.DSC"
        # Sample name: 100 uM TEMPO-SO4 in water aged prepared from solid 250110
    ),    
]:
    d[k] = find_file(fname, exp_type="francklab_esr/romana")
    d[k] = d[k].chunk_auto("harmonic", "harmonic")["harmonic", 0]["phase", 0]
    
```

```python
# TEMPO-OH
figure(dpi=300, figsize=(6,4)),
for k, thislabel in [
    ("d_TEMPO_OH_50mM", "50 mM TEMPO-OH in water"),
    ("d_TEMPO_OH_100uM_try1", "100 uM TEMPO-OH in water"),
]:
    d[k] = d[k]["$B_0$":(3400, 3550)]
    plot(
        d[k]/d[k].max(),
        label=thislabel,
        alpha=0.8,
        lw=0.9,
    )
gca().set_title("TEMPO-OH normalized"),
legend(
    bbox_to_anchor=(1.05, 1), loc=2, borderaxespad=0
)  # put the legend on the outside
```

```python
# 100 uM TEMPO-OH reproducibility check
# not-normalize
figure(dpi=300, figsize=(6,4)),
for k, thislabel in [
    ("d_TEMPO_OH_100uM_try1", "Trial 1"),
    ("d_TEMPO_OH_100uM_try2", "Trial 2"),
]:
    d[k] = d[k]["$B_0$":(3400, 3550)]
    plot(
        d[k],
        label=thislabel,
        alpha=0.8,
        lw=0.9,
    )
gca().set_title("100 uM TEMPO-OH"),
legend()

# Normalized
figure(dpi=300, figsize=(6,4)),
for k, thislabel in [
    ("d_TEMPO_OH_100uM_try1", "Trial 1"),
    ("d_TEMPO_OH_100uM_try2", "Trial 2"),
]:
    d[k] = d[k]["$B_0$":(3400, 3550)]
    plot(
        d[k]/d[k].max(),
        label=thislabel,
        alpha=0.8,
        lw=0.9,
    )
gca().set_title("100 uM TEMPO-OH normalized"),
legend()
```

```python
# TEMPO-SO4
figure(dpi=300, figsize=(6,4)),
for k, thislabel in [
    ("d_TEMPO_SO4_50mM", "50 mM TEMPO-SO$_4$ in water"),
    ("d_TEMPO_SO4_100uM_try1", "100 uM TEMPO-SO$_4$ in water"),
]:
    d[k] = d[k]["$B_0$":(3400, 3550)]
    plot(
        d[k]/d[k].max(),
        label=thislabel,
        alpha=0.8,
        lw=0.9,
    )
gca().set_title("TEMPO-SO$_4$ normalized"),
legend(
    bbox_to_anchor=(1.05, 1), loc=2, borderaxespad=0
)  # put the legend on the outside
```
