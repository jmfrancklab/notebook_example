```python
d1 = find_file("231121_20mMTEMPOSO4_W30.DSC", exp_type="francklab_esr/romana") #cooling low conc in isooctane
d2 = find_file("231116_TEMPOSO4_W30.DSC", exp_type="francklab_esr/romana") #heating subsequently high conc in isooctane
```

```python
d1.chunk_auto("harmonic", "harmonic")  # the harmonic is typically stored so that we need to do this
d1 = d1['phase',0]['harmonic',0]
d1.shape  # this shows me what the dimensions are
```

```python
d2 = d2['harmonic',0]
d2.shape
```

```python
#near room temperature
#print(d1.getaxis("Temperature"))
figure(dpi=300, figsize=(8,4))
plot(d1, label="20 mM $w_0= 30$ Isooctane dispersant")
plot(d2, label="100 mM $w_0= 30$ Isooctane dispersant")
title("TEMPO-SO$_4$ w$_0= 30$ in Isooctane at room temperature")
legend()
```

```python
#Normalized
figure(dpi=300, figsize=(7,4))
plot(d1/d1.max(), label= "20 mM $w_0= 30$ Isooctane dispersant")
plot(d2/d2.max(), label= "100 mM $w_0= 30$ Isooctane dispersant")
title("Normalized TEMPO-SO$_4$ w$_0= 30$ in Isooctane at room temperature)")
legend()
```
