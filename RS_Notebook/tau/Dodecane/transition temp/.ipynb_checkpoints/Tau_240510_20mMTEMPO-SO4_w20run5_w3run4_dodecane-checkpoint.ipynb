# Pick the Peaks and Calculate the Correlation times

The first cell here only operates over a limited temperature range.

```python
correlation_time_w3 = {}
```

```python
all_data= d_w3['Temperature':(None,None)] 
x_counter = 0
j_counter = 0
with figlist_var() as fl: 
    for this_temp in all_data['Temperature']:
        this_color = next(colors)
        label_str = str(this_temp)
        d = all_data["Temperature": this_temp]
        d /= d.max().item() 
        peak_data = {}
        # next, we define the ranges for our peaks
        # note that we can replace this with an if ... else statement
        # to allow different peak slices for different temperatures
        if this_temp >= 265:
            peak_A_range = (3315,3335)
            peak_B_range = (3335,3352)
            peak_C_range = (3351,3390)
#         elif this_temp >= 250:
#             peak_A_range = (3320,3358)
#             peak_B_range = (3358,3382)
#             peak_C_range = (3386,3403)
        else:
            peak_A_range = (3315,3335)
            peak_B_range = (3335,3353)
            peak_C_range = (3353,3385)
        for peak_range,peak_name in [
            (peak_A_range,'A'),
            (peak_B_range,'B'),
            (peak_C_range,'C'),
        ]:
            fl.next(f'slice peak {peak_name}') # move to the figure called "slice peak A", "slice peak B", etc.
            plt.title(fl.current) # this shouldn't be needed -- just says use the current figure name as the title
            peak_spline_fn = d['$B_0$':peak_range].spline_lambda() # create a function that will make a spline curve
            peak = peak_spline_fn(linspace(*tuple(peak_range+(500,)))) # make a smooth, high-res curve of our peak, using the spline
            plot(peak,c=this_color,alpha=0.2)
            peak = d['$B_0$':peak_range].spline_lambda(s_multiplier=2e-4)(linspace(*tuple(peak_range+(500,))))
            plot(peak,c=this_color,alpha=0.5) # overwrite with a smoothed one
            max_B0 = peak.C.argmax().item() # this one returns an nddata, while the argmin returns something that
            #                                 looks like a dictionary
            min_B0 = peak.C.argmin()['$B_0$']  # give the x axis values for the peaks -- there's a bug where
            #                                    this and the previous line should be similar, but are not
            # {{{ here we are builing up a dictionary that has all the values we need to calculate the correlation time
            peak_data[peak_name+'max'] = peak['$B_0$':max_B0].item() # return the y value, and store as value of key Amax, Bmax, Cmax
            peak_data[peak_name+'min'] = peak['$B_0$':min_B0].item()
            peak_data[peak_name+'maxB0'] = max_B0 # set the key AmaxB0, BmaxB0, etc. to max_B0
            peak_data[peak_name+'minB0'] = min_B0
            # }}}
            # {{{ explicitly check that the points that correspond to these values show up in the right place
            plot(peak_data[peak_name+'maxB0'],peak_data[peak_name+'max'],'o',c=this_color,alpha=0.5)
            plot(peak_data[peak_name+'minB0'],peak_data[peak_name+'min'],'x',c=this_color,alpha=0.5)
            # }}}
        fl.next('pub_fig',figsize=(8,12/(1+sqrt(5))*2), dpi=400) ## use golden ratio for figure size
        fl.plot(d+x_counter,alpha=0.8,color=this_color, human_units=False) ### x_counter offsets the curves
        # next, I have a dictionary that contains all the values I need to calculate my correlation time
        # this line plugs those values into the expression for my correlation time:
        # basically, it assumes that all of the words correspond to keys of my dictionary, and it substitutes
        # them with the corresponding values
        tau_c = eval("6.51e-10*(BminB0-BmaxB0)*(sqrt((Bmax-Bmin)/(Amax-Amin))+sqrt((Bmax-Bmin)/(Cmax-Cmin))-2)",{'sqrt':sqrt},peak_data)
        # above is equivalent to: 6.51e-10*(peak_data['BminB0']-peak_data['BmaxB0']*(sqrt((...
        print("correlation time is",tau_c)
        correlation_time_w3[label_str] = tau_c 
        plt.gca().text(
                3370,
                x_counter,
                label_str+f' K\n$\\tau_c={tau_c/1e-9:0.2f}$ ns',
                transform=plt.gca().transData,
                va='bottom',
                ha='left',
                color=this_color,
                )
        x_counter += 1
        #xlim(3375,3550)
        j_counter += 1.2
    fl.next('pub_fig')
    plt.title('')
    plt.ylabel('')
    plt.gca().tick_params(left=False, bottom=True, right=False, top=False,
                          labelleft=False, labelbottom=True,
                          labelright=False)
```

```python
correlation_time_w3
```

```python
temp_w3= np.array([float(j) for j in correlation_time_w3.keys()])
tau_w3 = np.array(list(correlation_time_w3.values()))
```

```python
all_data_w20= d_w20['Temperature':(None,None)] # let's start just with high temp
x_counter = 0
j_counter = 0
with figlist_var() as fl: ### why with???
    for this_temp in all_data_w20['Temperature']:
        this_color = next(colors)
        label_str = str(this_temp)
        d = all_data_w20["Temperature": this_temp]
        d /= d.max().item() ### RS item??? normalize??
        peak_data_w20 = {}
        # next, we define the ranges for our peaks
        # note that we can replace this with an if ... else statement
        # to allow different peak slices for different temperatures
        if this_temp >= 230:
            peak_A_range = (3320,3335)
            peak_B_range = (3335,3355)
            peak_C_range = (3355,3375)
#         elif this_temp >= 250:
#             peak_A_range = (3320,3358)
#             peak_B_range = (3358,3382)
#             peak_C_range = (3376,3403)            
#         else:
#             peak_A_range = (3320,3358)
#             peak_B_range = (3358,3382)
#             peak_C_range = (3382,3406)
        for peak_range,peak_name in [
            (peak_A_range,'A'),
            (peak_B_range,'B'),
            (peak_C_range,'C'),
        ]:
            fl.next(f'slice peak {peak_name}') # move to the figure called "slice peak A", "slice peak B", etc.
            plt.title(fl.current) # this shouldn't be needed -- just says use the current figure name as the title
            peak_spline_fn = d['$B_0$':peak_range].spline_lambda() # create a function that will make a spline curve
            peak = peak_spline_fn(linspace(*tuple(peak_range+(500,)))) # make a smooth, high-res curve of our peak, using the spline
            plot(peak,c=this_color,alpha=0.2)
            peak = d['$B_0$':peak_range].spline_lambda(s_multiplier=1e-4)(linspace(*tuple(peak_range+(500,))))
            plot(peak,c=this_color,alpha=0.5) # overwrite with a smoothed one
            max_B0 = peak.C.argmax().item() # this one returns an nddata, while the argmin returns something that
            #                                 looks like a dictionary
            min_B0 = peak.C.argmin()['$B_0$']  # give the x axis values for the peaks -- there's a bug where
            #                                    this and the previous line should be similar, but are not
            # {{{ here we are builing up a dictionary that has all the values we need to calculate the correlation time
            peak_data_w20[peak_name+'max'] = peak['$B_0$':max_B0].item() # return the y value, and store as value of key Amax, Bmax, Cmax
            peak_data_w20[peak_name+'min'] = peak['$B_0$':min_B0].item()
            peak_data_w20[peak_name+'maxB0'] = max_B0 # set the key AmaxB0, BmaxB0, etc. to max_B0
            peak_data_w20[peak_name+'minB0'] = min_B0
            # }}}
            # {{{ explicitly check that the points that correspond to these values show up in the right place
            plot(peak_data_w20[peak_name+'maxB0'],peak_data_w20[peak_name+'max'],'o',c=this_color,alpha=0.5)
            plot(peak_data_w20[peak_name+'minB0'],peak_data_w20[peak_name+'min'],'x',c=this_color,alpha=0.5)
            # }}}
        fl.next('pub_fig',figsize=(10,10/(1+sqrt(5))*2), dpi=400) ## use golden ratio for figure size
        fl.plot(d+x_counter,alpha=0.8,color=this_color, human_units=False) ### x_counter offsets the curves
        # next, I have a dictionary that contains all the values I need to calculate my correlation time
        # this line plugs those values into the expression for my correlation time:
        # basically, it assumes that all of the words correspond to keys of my dictionary, and it substitutes
        # them with the corresponding values
        tau_c_w20 = eval("6.51e-10*(BminB0-BmaxB0)*(sqrt((Bmax-Bmin)/(Amax-Amin))+sqrt((Bmax-Bmin)/(Cmax-Cmin))-2)",{'sqrt':sqrt},peak_data_w20)
        # above is equivalent to: 6.51e-10*(peak_data['BminB0']-peak_data['BmaxB0']*(sqrt((...
        print("correlation time is",tau_c)
        correlation_time_w20[label_str] = tau_c_w20 
        plt.gca().text(
                3370,
                x_counter,
                label_str+f' K\n$\\tau_c={tau_c_w20/1e-9:0.1f}$ ns',
                transform=plt.gca().transData,
                va='bottom',
                ha='left',
                color=this_color,
                )
        x_counter += 1.5
        #xlim(3375,3550)
        j_counter += 1
    fl.next('pub_fig')
    plt.title('')
    plt.ylabel('')
    plt.gca().tick_params(left=False, bottom=True, right=False, top=False,
                          labelleft=False, labelbottom=True,
                          labelright=False)
```

```python
# Tau caculation
figure(dpi=300, figsize=(6,4))
plot(temp_w3, tau_w3/1e-9, linestyle='-',  marker='o', label=r"$\tau_c$ for $w_0=3$")
plot(temp_w20, tau_w20/1e-9, linestyle='-',  marker='o', label=r"$\tau_c$ for $w_0=20$")
legend()
xlabel('Temperature (K)')
ylabel('Correlation Time (ns)')
title('Correlation Time from linewidth for 20 mM TEMPO-SO$_4$ in dodecane')
show()
```

```python
# Arrhenius  
figure(dpi=300, figsize=(6,4))
plot(1000/temp_w3, log10(tau_w3*temp_w3), linestyle='-',  marker='o', label=r"for $w_0=3$")
plot(1000/temp_w20, log10(tau_w20*temp_w20), linestyle='-',  marker='o', label=r"for $w_0=20$")
legend()
# xlim(2.9,4.1)
xlabel(r"1000/Temperature $(K^{-1})$")
ylabel(r"$logT+ log(\tau_c)$")
title('Arrhenius temperature dependance from linewidth')
show()
```

# Correlation time from Multicomponent fitting

```python
# Sample name: 	20 mM TEMPO-SO4 in Dodecane $w_0=3$
R1_w3 = {
'223': 7.670897,
'253.169': 7.935684,
'256.653': 7.95353,
'261.649': 7.975954,
'268.351': 8.011591,
'274.489': 8.045357,
'282.183': 8.10043,
'293': 7.571283}
```

```python
# Sample name: 	20 mM TEMPO-SO4 in Dodecane $w_0=20$
R1_w20 = {
'243.977': 8.161657,
'249.372': 8.161175,
'254.425': 8.158947,
'261.322': 8.161238,
'270.679': 8.160712,
'283.363': 8.157241}
```

```python
temp_fit_w3= np.array([float(j) for j in R1_w3.keys()])
R1_fit_w3 = np.array(list(R1_w3.values()))
temp_fit_w20= np.array([float(j) for j in R1_w20.keys()])
R1_fit_w20 = np.array(list(R1_w20.values()))
```

```python
tau_fit_w3 = 0.5241/(pi*10**R1_fit_w3)
tau_fit_w20 = 0.5241/(pi*10**R1_fit_w20)
```

```python
print(tau_fit_w3)
print(tau_fit_w20)
```

```python
# Tau plot from fitting
figure(dpi=300, figsize=(6,4))
plot(temp_fit_w3, tau_fit_w3/1e-9, linestyle='-',  marker='o', label=r"$\tau_c$ for $w_0=3$")
plot(temp_fit_w20, tau_fit_w20/1e-9, linestyle='-',  marker='o', label=r"$\tau_c$ for $w_0=20$")
legend()
xlabel('Temperature (K)')
ylabel('Correlation Time (ns)')
title('Correlation Time from fitting for 20 mM TEMPO-SO$_4$ in dodecane')
show()
```

```python
# Arrhenius Temperature dependance 
figure(dpi=300, figsize=(6,5))

# from linewidth
plot(1000/temp_w3, log10(tau_w3*temp_w3), linestyle='-',  label=r"for $w_0=3$ from linewidth")
plot(1000/temp_w20, log10(tau_w20*temp_w20), linestyle='-',  label=r"for $w_0=20$ from linewidth")

# from fitting
plot(1000/temp_fit_w3, log10(tau_fit_w3*temp_fit_w3), marker='o', linestyle='', label=r"for $w_0=3$ from fitting")
plot(1000/temp_fit_w20, log10(tau_fit_w20*temp_fit_w20), marker='x', linestyle='',  label=r"for $w_0=20$ from fitting")
legend()
xlabel(r"1000/Temperature $(K^{-1})$")
ylabel(r"$logT+ log(\tau_c)$")
title('Arrhenius temperature dependance for 20 mM TEMPO-SO$_4$ in dodecane')
show()
```
