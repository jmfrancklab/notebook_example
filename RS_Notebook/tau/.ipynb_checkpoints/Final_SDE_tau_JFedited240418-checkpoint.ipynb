```python
%load_ext pyspecdata.ipy
import numpy as np
from numpy import r_
from pyspecdata import k_B
import scipy
```

these are pulled from code I gave you earlier -- the point here is just to get the rm radius

equations from Pal2011FluCorSpe

```python
def smaller_radius(w):
    return 0.175*w+1.5
def larger_radius(w):
    delta = 1.5
    gamma = 0.043
    y = 20.4
    numerator = y*(1+2*gamma)
    radius = 3*delta*w/numerator*(
            1+numerator/3/w
            -numerator**2/9/w**2
            +5*numerator**3/81/w**3)
    return radius
@np.vectorize
def rm_radius(w):
    return smaller_radius(w) if w<20 else larger_radius(w)
```

```python
print(rm_radius(3))
```

I would suggest going something like this.  Remember that we're going to want to calculate
$$\tau T  = \frac{4 \pi R^3 \eta}{3 k_B}$$
`tau_T_SDE_w3` below gives the $\tau T$ value predicted by SDE for each temperature (listed in the same order as the temperatures)
To get numbers below, we note that 1 cP = 1e-3 Pa·s (with the former being common and the latter being SI)
I also need to convert the radius nm = 1e-9 m

This is reassuring because we can see the math that's use to calculate right here, and we can agree or disagree on the math
used to generate these numbers.

```python
T_for_viscosity = r_[323.16, 273, 222.91] # r_ is an easy way of constructing numpy arrays
viscosity = r_[0.365e-3, 0.65e-3, 1.58e-3] 
tau_T_SDE_w3 = 4 * pi * (rm_radius(3)*1e-9)**3 * viscosity / (3 * k_B)
tau_T_SDE_w7 = 4 * pi * (rm_radius(7)*1e-9)**3 * viscosity / (3 * k_B)
tau_T_SDE_w20 = 4 * pi * (rm_radius(20)*1e-9)**3 * viscosity / (3 * k_B)
```

<font color="red">RS: I convert your dictionaries to numpy arrays.
Note that since these are manually entered, it would have been just as easy to enter them as I'm suggesting for temperatures and viscosities above.
The general way that you do this is to match the temperature and the value at the same index -- in that way, they are
both used together.</font>

(We could do this by collecting both into a single nddata, too, of course, but I figure it's good to learn the pure numpy way as well.)

```python
temp_w3 = np.array([float(j) for j in correlation_time_w3.keys()])
tau_w3 = np.array(list(correlation_time_w3.values()))
temp_w20 = np.array([float(j) for j in correlation_time_w20.keys()])
tau_w20 = np.array(list(correlation_time_w20.values()))
```

Here, I generate the Arrhenius plot right away.  Note that what I'm doing below (element-wise division and logarithm) is only possible with numpy -- it's not possible with lists.  Lists don't know how to do element-by-element operations automatically like this -- only numpy.

Previously, the code was build up step-wise, which is fine, but rather than repeating, I just put it all into one plot right away

```python
# Arrhenius
figure(dpi=300, figsize=(6,4))
plot(1000/temp_w3, log10(tau_w3*temp_w3), linestyle='-',  marker='o', label=r"from ESR, $w_0=3$")
plot(1000/temp_w20, log10(tau_w20*temp_w20), linestyle='-',  marker='o', label=r"from ESR, $w_0=20$")
plot(1000/T_for_viscosity, log10(tau_T_SDE_w3), linestyle='-',  marker='o', alpha=0.5, label=r"$\tau_{RM}T$"+"\n(from SDE based on $w_0=3$ radius)")
plot(1000/T_for_viscosity, log10(tau_T_SDE_w7), linestyle='-',  marker='o', alpha=0.5, label=r"$\tau_{RM}T$"+"\n(from SDE based on $w_0=7$ radius)")
plot(1000/T_for_viscosity, log10(tau_T_SDE_w20), linestyle='-',  marker='o', alpha=0.5, label=r"$\tau_{RM}T$"+"\n(from SDE based on $w_0=20$ radius)")
legend(bbox_to_anchor=(1.05,1), loc=2, borderaxespad=0.) # fancy_legend() is just an abbreviation to this -- you can add other arguments as needed now
xlabel('1000/($T$ / K)')
ylabel(r'$\log(\tau T / \mathrm{s} \cdot \mathrm{K})$')
```
