## Isooctane TEMPO-SO4 W3 

```python
correlation_time_cooling = {'213.565': 1.4684243780660179e-08,
 '223.612': 1.3932821555987277e-08,
 '233.819': 1.4897588829708854e-08,
 '243.863': 1.8244270500325767e-08,
 '252.461': 1.500406705961571e-08, 
 '262.57': 6.153557611053051e-09,
 '272.297': 3.2257053012370264e-09,
 '282.495': 2.006856871875143e-09,
 '292.354': 1.304493313755594e-09,
 '302.071': 8.577447785613093e-10}
```

```python
correlation_time_reheating = {'212.999': 1.4060349948362108e-08,
 '223.344': 1.4119168631492322e-08,
 '233.419': 1.6222108521419707e-08,
 '243.593': 2.1298362772228006e-08,
 '253.269': 1.2792813220185145e-08,
 '263.403': 5.7071750435811634e-09,
 '273.454': 3.1137398795973714e-09,
 '283.306': 1.987413952502945e-09,
 '293.575': 1.240241444638978e-09,
 '302.696': 8.172671735144214e-10}
```

```python
temp_cooling= np.array([float(j) for j in correlation_time_cooling.keys()])
tau_cooling = np.array(list(correlation_time_cooling.values()))
temp_reheating= np.array([float(j) for j in correlation_time_reheating.keys()])
tau_reheating = np.array(list(correlation_time_reheating.values()))
```

```python
# Tau caculation
figure(dpi=300, figsize=(6,4))
plot(temp_cooling[3:], tau_cooling[3:]/1e-9, linestyle='-',  marker='o', label=r"$\tau_c$ for cooling")
plot(temp_reheating[3:], tau_reheating[3:]/1e-9, linestyle='-',  marker='o', label=r"$\tau_c$ for reheating")
legend()
xlabel('Temperature (K)')
ylabel('Correlation Time (ns)')
title('Correlation Time for TEMPO-SO$_4$ w$_0=3$')
show()
```

```python
# Arrhenius  
figure(dpi=300, figsize=(6,4))
plot(1000/temp_cooling[3:], log10(tau_cooling[3:]*temp_cooling[3:]), linestyle='-',  marker='o', label=r"for cooling")
plot(1000/temp_reheating[3:], log10(tau_reheating[3:]*temp_reheating[3:]), linestyle='-',  marker='o', label=r"for reheating")
legend()
xlabel(r"1000/Temperature $(1/K)$")
ylabel(r"$log(\tau_cT)$")
title('Arrhenius Temperature dependance for TEMPO-SO$_4$ w$_0=3$ in isooctane')
show()
```
