```python
# load file
# d_VT = {}  # dictionary of variable temperature ESR nddata
for k, fname in [
    # 20 mM TEMPO-SO4 in isooctane $w_0=20$
    (
        "d_TEMPOSO4_w20_250130_VT",
        "250130_20mM_TSO4_W20_Isooct_run1.DSC",
        # Sample name: equlibrate 1 hour
    ),
     # 20 mM TEMPO-SO4 in isooctane $w_0=3$
    (
        "d_TEMPOSO4_w3_250128",
        "250128_20mM_TSO4_W3_Isooct.DSC",
        # Sample name: water loading 3
    )
]:
    d_VT[k] = find_file(fname, exp_type="francklab_esr/romana")
    d_VT[k] = d_VT[k].chunk_auto("harmonic", "harmonic")["harmonic", 0]["phase", 0]
    d_VT[k].reorder('Temperature').sort('Temperature')

# d_RT = {} # dictionary of room temperature ESR nddata
for k, fname in [
    # 20 mM TEMPO-SO4 in isooctane $w_0=20$
    (
        "d_TEMPOSO4_W20_250131",
        "250131_20mM_TSO4_Isooct_w20_prep250130.DSC",
        # Sample name: equlibrate 1 day. prepared 250130
    ),
    (
        "d_TEMPOSO4_W20_250131_after_cooling_bath",
        "250131_20mM_TSO4_Isooct_w20_prep250130_2.DSC",
        # Sample name: equlibrate 1 day. After cooling bath
    ),
    (
        "d_TEMPOSO4_W20_250131_after_cooling_bath2",
        "250131_20mM_TSO4_Isooct_w20_prep250130_3.DSC",
        # Sample name: equlibrate 1 day. After cooling bath
    ),
    # 20 mM TEMPO-SO4 in isooctane $w_0=3$
    (
        "d_TEMPOSO4_w3_250131",
        "250131_20mM_TSO4_Isooct_w3_prep250128.DSC",
        # Sample name: water loading 3
    )
]:
    d_RT[k] = find_file(fname, exp_type="francklab_esr/romana")
    d_RT[k] = d_RT[k].chunk_auto("harmonic", "harmonic")["harmonic", 0]["phase", 0]
```

```python
# TEMPO-SO4 20 mM w0=20 cooling bath 
figure(dpi=300, figsize=(6,4)),
for k, thislabel in [
    ("d_TEMPOSO4_W20_250131", "before cooling bath"),
    ("d_TEMPOSO4_W20_250131_after_cooling_bath2", "after cooling bath"),
    ("d_TEMPOSO4_W20_250131_after_cooling_bath2", "after cooling bath more"),
]:
    d_RT[k] = d_RT[k]["$B_0$":(3300, 3550)]
    plot(
        d_RT[k],
        label=thislabel,
        alpha=0.8,
        lw=0.9,
    )
gca().set_title("20 mM TEMPO-SO$_4$ $w_0=20 before and after cooling bath"),
legend(bbox_to_anchor=(1.05, 1), loc=2, borderaxespad=0)
show()

# normalized
figure(dpi=300, figsize=(6,4)),
for k, thislabel in [
    ("d_TEMPOSO4_W20_250131", "before cooling bath"),
    ("d_TEMPOSO4_W20_250131_after_cooling_bath2", "after cooling bath"),
    ("d_TEMPOSO4_W20_250131_after_cooling_bath2", "after cooling bath more"),
]:
    d_RT[k] = d_RT[k]["$B_0$":(3300, 3550)]
    plot(
        d_RT[k]/d_RT[k].max(),
        label=thislabel,
        alpha=0.8,
        lw=0.9,
    )
gca().set_title("20 mM TEMPO-SO$_4$ $w_0=20 before and after cooling bath"),
legend(bbox_to_anchor=(1.05, 1), loc=2, borderaxespad=0)
show()
```
